Objectif : transcrire les extraits audios en texte

```python
import whisper
import os

# Configuration
TARGET_DURATION = 7.0  # Durée cible pour chaque "scène" en secondes
model = whisper.load_model("base")
input_folder = "../Audios/Original_files"
output_base_folder = "../Transcriptions"

os.makedirs(output_base_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.endswith(".mp3"):
        audio_path = os.path.join(input_folder, filename)
        # On crée un sous-dossier par audio pour ne pas mélanger les parties
        audio_name = os.path.splitext(filename)[0]
        audio_output_dir = os.path.join(output_base_folder, audio_name)
        os.makedirs(audio_output_dir, exist_ok=True)

        print(f"⏳ Analyse de {filename}...")
        result = model.transcribe(audio_path, verbose=False)
        
        current_part_text = ""
        current_part_start = 0
        part_index = 1

        # Whisper renvoie une liste de 'segments' avec start, end et text
        for segment in result["segments"]:
            current_part_text += segment["text"] + " "
            
            # Si le segment dépasse la durée voulue, on sauvegarde
            if segment["end"] - current_part_start >= TARGET_DURATION:
                part_filename = f"{audio_name}_part{part_index:02d}.txt"
                with open(os.path.join(audio_output_dir, part_filename), "w", encoding="utf-8") as f:
                    f.write(current_part_text.strip())
                
                print(f"   ✅ Partie {part_index} générée ({segment['end']:.1f}s)")
                
                # Réinitialisation pour la partie suivante
                current_part_text = ""
                current_part_start = segment["end"]
                part_index += 1

        # Sauvegarde du dernier morceau s'il reste du texte
        if current_part_text.strip():
            part_filename = f"{audio_name}_part{part_index:02d}.txt"
            with open(os.path.join(audio_output_dir, part_filename), "w", encoding="utf-8") as f:
                f.write(current_part_text.strip())

        print(f"🏁 Terminé ! {part_index} parties créées pour {filename}.\n")
```
